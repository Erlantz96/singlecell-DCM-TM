```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Set matplotlib parameters for vector-friendly, publication-ready figures
plt.rcParams['pdf.fonttype'] = 42  # TrueType fonts (editable text in PDF)
plt.rcParams['ps.fonttype'] = 42   # TrueType fonts for PS/EPS
plt.rcParams['savefig.dpi'] = 300

celltypes = adata_subset2_cy.obs['leiden_annot_25_CR']
try:
    lineage_names = adata_subset2_cy.obsm['lineages_fwd'].names
except AttributeError:
    lineage_names = [f'Lineage_{i+1}' for i in range(adata_subset2_cy.obsm['lineages_fwd'].shape[1])]

# Create the DataFrame
lineages_fwd = pd.DataFrame(
    adata_subset2_cy.obsm['lineages_fwd'], 
    index=adata_subset2_cy.obs.index,
    columns=lineage_names
)

# Combine both into a single DataFrame
combined_df = pd.concat([celltypes, lineages_fwd], axis=1)
aggregated_df = combined_df.groupby('leiden_annot_25_CR').agg('mean')

# Normalize by the number of samples per cell type
sample_counts = combined_df['leiden_annot_25_CR'].value_counts()
normalized_df = aggregated_df.div(sample_counts, axis=0)

# Normalize each column so that the sum of each column is 1
probability_df = normalized_df.div(normalized_df.sum(axis=1), axis=0)
grouped_columns = probability_df.columns.str.replace(r'_\d+$', '', regex=True)

# Sum the probabilities within each group
summed_probabilities = probability_df.groupby(grouped_columns, axis=1).sum()

# Define the desired orders
main_cell_types = ["Stem", "TA", "Enterocyte", "Paneth", "Goblet", "Tuft", "EEC"]

desired_ordery = ["Stem", "TA", "Enterocyte", "Paneth", "Goblet", "Tuft", "EEC"]

# Color palette
color_palette = ["#9370DB", "#D2B48C", "#FF8C00", "#DC143C", "#2E8B57", "#FF1493", '#4169E1']
cell_type_colors = dict(zip(["Stem", "TA", "Enterocyte", "Paneth", "Goblet", "Tuft", "EEC"], color_palette))

# Cell type annotations from the table
cell_annotations = {
    "Stem_Batch_1_Plate_2_D06": ("Stem", "Stem"),
    "Enterocyte_Batch_1_Plate_2_K06": ("Stem", "Enterocyte"),
    "Stem_Batch_1_Plate_2_B07": ("Stem", "Stem"),
    "Tuft_Batch_1_Plate_1_E18": ("Stem", "Tuft"),
    "Tuft_Batch_1_Plate_1_B17": ("Stem", "Tuft"),
    "Stem_Batch_1_Plate_2_H07": ("Stem", "Stem"),
    "EEC_Batch_1_Plate_1_A07": ("Stem", "EEC"),
    "EEC_Batch_1_Plate_1_B19": ("Stem", "EEC"),
    "Tuft_Batch_2_Plate_2_C22": ("Stem", "Tuft"),
    "Enterocyte_Batch_2_Plate_3_O22": ("Stem", "Enterocyte"),
    "Stem_Batch_2_Plate_3_A21": ("Stem", "Stem"),
    "Stem_Batch_2_Plate_3_H07": ("Stem", "Stem"),
    "Enterocyte_Batch_2_Plate_3_M16": ("Stem", "Enterocyte"),
    "EEC_Batch_2_Plate_2_F24": ("Stem", "EEC"),
    "Stem_Batch_2_Plate_3_H12": ("Stem", "Stem"),
    "Stem_Batch_2_Plate_3_F11": ("Stem", "Stem"),
    "Paneth_Batch_1_Plate_1_L09": ("TA", "Paneth"),
    "Stem_Batch_1_Plate_2_H11": ("TA", "Stem"),
    "Enterocyte_Batch_1_Plate_2_J07": ("TA", "Enterocyte"),
    "Paneth_Batch_1_Plate_1_I23": ("TA", "Paneth"),
    "EEC_Batch_1_Plate_1_B23": ("TA", "EEC"),
    "EEC_Batch_1_Plate_1_B21": ("TA", "EEC"),
    "Enterocyte_Batch_1_Plate_2_C06": ("TA", "Enterocyte"),
    "Enterocyte_Batch_1_Plate_2_J24": ("TA", "Enterocyte"),
    "TA_Batch_1_Plate_2_B20": ("TA", "TA"),
    "TA_Batch_1_Plate_1_G22": ("TA", "TA"),
    "Stem_Batch_2_Plate_3_D23": ("TA", "Stem"),
    "Goblet_Batch_2_Plate_3_C22": ("TA", "Goblet"),
    "TA_Batch_2_Plate_3_G09": ("TA", "TA"),
    "TA_Batch_2_Plate_3_B12": ("TA", "TA"),
    "Goblet_Batch_2_Plate_2_O17": ("TA", "Goblet"),
    "TA_Batch_2_Plate_3_H24": ("TA", "TA"),
    "Tuft_Batch_2_Plate_2_M23": ("TA", "Tuft")
}

# Order individual cells by RNA type first, then by DCM type within each RNA group
dcm_order = ["Stem", "TA", "Enterocyte", "Paneth", "Goblet", "Tuft", "EEC"]

# Sort cells: first by RNA type, then by DCM type
def get_sort_key(cell_id):
    dcm_type, rna_type = cell_annotations[cell_id]
    dcm_rank = dcm_order.index(dcm_type)
    rna_rank = main_cell_types.index(rna_type)
    return (dcm_rank, rna_rank)

individual_cells_sorted = sorted(cell_annotations.keys(), key=get_sort_key)

# ============================================
# SPLIT 1: Main Cell Types only
# ============================================
summed_probabilities_main = summed_probabilities.reindex(index=main_cell_types)
summed_probabilities_main = summed_probabilities_main.reindex(columns=desired_ordery)

# Plot 1: Main Cell Types Heatmap
fig1, ax1 = plt.subplots(figsize=(6, 4.5))
sns.heatmap(
    summed_probabilities_main.T, 
    annot=False,
    cmap="RdBu_r",
    cbar_kws={'label': 'Probability'},
    vmin=0, vmax=1,
    ax=ax1
)

# Add colored bar annotations ABOVE the heatmap (negative Y position)
for i, cell_type in enumerate(main_cell_types):
    ax1.add_patch(plt.Rectangle((i, -0.7), 1, 0.5, 
                                facecolor=cell_type_colors[cell_type], 
                                edgecolor='black', linewidth=1, clip_on=False))

ax1.set_title("Cytotrace Probabilities - Main Cell Types", fontsize=16, weight='bold', pad=15)
ax1.set_xlabel("Source Cell Type", fontsize=14, weight='bold')
ax1.set_ylabel("Destination Cell Type", fontsize=14, weight='bold')
ax1.tick_params(axis='both', which='major', labelsize=12)

plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.savefig('/media/erlantz/71DE167A75E99BB4/Pool40_rebuttal/Jona_py/CellRank/Cytotrace_kernel/probability_heatmap_main_celltypes_poscontrol.pdf', 
            dpi=300, bbox_inches='tight')
plt.show()

# ============================================
# SPLIT 2: Individual Cells
# ============================================
summed_probabilities_individuals = summed_probabilities.reindex(index=individual_cells_sorted)
summed_probabilities_individuals = summed_probabilities_individuals.reindex(columns=desired_ordery)

# Plot 2: Individual Cells Heatmap
fig2, ax2 = plt.subplots(figsize=(12, 6.5))
sns.heatmap(
    summed_probabilities_individuals.T, 
    annot=False,
    cmap="RdBu_r",
    cbar_kws={'label': 'Probability'},
    vmin=0, vmax=1,
    ax=ax2
)

# Add two colored bars ABOVE the heatmap for DCM and RNA cell types with space in between
for i, cell_id in enumerate(individual_cells_sorted):
    dcm_type, rna_type = cell_annotations[cell_id]
    # DCM annotation (top bar - first row) - higher up
    ax2.add_patch(plt.Rectangle((i, -1.4), 1, 0.5, 
                                facecolor=cell_type_colors[dcm_type], 
                                edgecolor='black', linewidth=0.5, clip_on=False))
    # RNA annotation (second bar - second row) - closer to heatmap
    ax2.add_patch(plt.Rectangle((i, -0.7), 1, 0.5, 
                                facecolor=cell_type_colors[rna_type], 
                                edgecolor='black', linewidth=0.5, clip_on=False))

# Add text labels for DCM and RNA on the left side
ax2.text(-1.5, -1.15, 'DCM', fontsize=10, weight='bold', va='center', ha='right')
ax2.text(-1.5, -0.45, 'RNA', fontsize=10, weight='bold', va='center', ha='right')

ax2.set_title("Cytotrace Probabilities - Individual Cells", fontsize=16, weight='bold', pad=15)
ax2.set_xlabel("Source Individual Cells", fontsize=14, weight='bold')
ax2.set_ylabel("Destination Cell Type", fontsize=14, weight='bold')
ax2.tick_params(axis='both', which='major', labelsize=10)

plt.xticks(rotation=90, ha='right', fontsize=8)
plt.yticks(rotation=0)
plt.tight_layout()
plt.savefig('/media/erlantz/71DE167A75E99BB4/Pool40_rebuttal/Jona_py/CellRank/Cytotrace_kernel/probability_heatmap_individual_cells_poscontrol.pdf', 
            dpi=300, bbox_inches='tight')
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from io import StringIO

# ─────────────────────────────────────────────
# DATA
# Paste your data here as a tab-separated string,
# or replace with: df = pd.read_csv('your_file.csv')
# ─────────────────────────────────────────────
df = pd.read_csv("/media/erlantz/71DE167A75E99BB4/Pool40_rebuttal/Jona_py/CellRank/Nature/Combi_kernel/probabilities_stable_unstable.csv", sep=',')


# ─────────────────────────────────────────────
# COMPUTE  p_other = 1 − P(own cell type)
# ─────────────────────────────────────────────
cell_type_cols = ['Stem', 'TA', 'Enterocyte', 'Paneth', 'Goblet', 'Tuft', 'EEC']

# For each row look up the probability in the column that matches CellType
df['p_self']  = df.apply(lambda r: r[r['CellType']], axis=1)
df['p_other'] = 1 - df['p_self']

print("Summary of p_other by Transition:")
print(df.groupby('Transition')['p_other'].describe().round(3))
print()
print("Summary of p_other by CellType × Transition:")
print(df.groupby(['CellType', 'Transition'])['p_other']
        .agg(['count', 'mean', 'median', 'std'])
        .round(3))

# ─────────────────────────────────────────────
# COLOURS
# ─────────────────────────────────────────────
cell_type_colors = {
    'Stem':       '#9370DB',
    'TA':         '#D2B48C',
    'Enterocyte': '#FF8C00',
    'Paneth':     '#DC143C',
    'Goblet':     '#2E8B57',
    'Tuft':       '#FF1493',
    'EEC':        '#4169E1',
}
transition_colors = {
    'Stable_Ctrl': '#95a5a6',   # gray  — negative control
    'Stable':      '#2ecc71',   # green
    'Unstable':    '#e74c3c',   # red
}
label_order = ['Stem', 'TA', 'Enterocyte', 'Paneth', 'Goblet', 'Tuft', 'EEC']

# ─────────────────────────────────────────────
# SIGNIFICANCE HELPER
# ─────────────────────────────────────────────
def get_sig(p):
    if pd.isna(p):   return ''
    if p < 0.001:    return '***'
    if p < 0.01:     return '**'
    if p < 0.05:     return '*'
    return 'ns'

# ─────────────────────────────────────────────
# PLOT 1: COMBINED — all cell types pooled
#   Stable vs Unstable, single boxplot
# ─────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(6, 6))

order = ['Stable_Ctrl', 'Stable', 'Unstable']
sns.boxplot(
    data=df, x='Transition', y='p_other',
    hue='Transition', order=order,
    palette=transition_colors, legend=False, ax=ax,
    flierprops=dict(marker='o', markersize=3, alpha=0.5),
    width=0.5,
)
sns.stripplot(
    data=df, x='Transition', y='p_other',
    hue='CellType', order=order,
    palette=cell_type_colors, dodge=False,
    alpha=0.7, size=5, jitter=True, ax=ax,
)
h, l = ax.get_legend_handles_labels()
ax.legend(h, l, title='Cell type', bbox_to_anchor=(1.01, 1),
          loc='upper left', fontsize=8, frameon=True)

# n labels + t-test bracket
ylim  = ax.get_ylim()
y_top = ylim[1]
for i, grp in enumerate(order):
    n = len(df[df['Transition'] == grp])
    ax.text(i, y_top * 1.01, f'n={n}',
            ha='center', va='bottom', fontsize=9, fontweight='bold')

# Pairwise t-tests: each group vs Stable_Ctrl
present_order = [g for g in order if g in df['Transition'].unique()]
ctrl_vals = df[df['Transition'] == 'Stable_Ctrl']['p_other'] if 'Stable_Ctrl' in df['Transition'].values else None

bracket_pairs = [
    ('Stable_Ctrl', 'Stable',   '#2ecc71'),
    ('Stable_Ctrl', 'Unstable', '#e74c3c'),
    ('Stable',      'Unstable', '#e74c3c'),
]
y_levels = [y_top * 1.10, y_top * 1.20, y_top * 1.30]
y_max_used = y_top

for (g1, g2, color), y_lvl in zip(bracket_pairs, y_levels):
    if g1 not in present_order or g2 not in present_order:
        continue
    v1 = df[df['Transition'] == g1]['p_other']
    v2 = df[df['Transition'] == g2]['p_other']
    if len(v1) < 2 or len(v2) < 2:
        continue
    _, pval = stats.ttest_ind(v2, v1)
    stars   = get_sig(pval)
    i1, i2  = present_order.index(g1), present_order.index(g2)
    ax.plot([i1, i1, i2, i2],
            [y_lvl - y_top*0.02, y_lvl, y_lvl, y_lvl - y_top*0.02],
            color='#333333', lw=1.2)
    pval_txt = f'p={pval:.2e}' if pval < 0.001 else f'p={pval:.3f}'
    ax.text((i1 + i2) / 2, y_lvl + y_top*0.01,
            f'{stars}  {pval_txt}',
            ha='center', va='bottom', fontsize=8, fontweight='bold',
            color='black' if stars != 'ns' else '#888888')
    y_max_used = max(y_max_used, y_lvl)

ax.set_ylim(ylim[0], y_max_used * 1.12)

ax.set_xlabel('')
ax.set_ylabel('P(transition away from own identity)\n= 1 − P(self)', fontsize=10)
ax.set_title('Probability of transitioning away\nfrom own identity: Stable_Ctrl vs Stable vs Unstable',
             fontsize=12, fontweight='bold')
ax.axhline(0.5, color='gray', lw=1, ls='--', alpha=0.6, label='0.5 threshold')
plt.tight_layout()
plt.savefig('p_other_combined2.pdf', dpi=300, bbox_inches='tight')
plt.show()
plt.close()
print("Saved: p_other_combined.pdf")

# ─────────────────────────────────────────────
# PLOT 2: PER CELL TYPE — one subplot each
#   Stable vs Unstable within each cell type
#   (only cell types present in both groups)
# ─────────────────────────────────────────────
ct_with_both = [
    ct for ct in label_order
    if set(df[df['CellType'] == ct]['Transition'].unique()) >= {'Stable', 'Unstable'}
]
ct_stable_only = [
    ct for ct in label_order
    if ct in df['CellType'].unique()
    and 'Unstable' not in df[df['CellType'] == ct]['Transition'].values
]
ct_unstable_only = [
    ct for ct in label_order
    if ct in df['CellType'].unique()
    and 'Stable' not in df[df['CellType'] == ct]['Transition'].values
]

print(f"\nCell types with both Stable & Unstable: {ct_with_both}")
print(f"Stable only : {ct_stable_only}")
print(f"Unstable only: {ct_unstable_only}")

# Show all cell types present; single group shown as single box
all_ct_present = [ct for ct in label_order if ct in df['CellType'].unique()]
n_ct = len(all_ct_present)

fig, axes = plt.subplots(1, n_ct, figsize=(4 * n_ct, 6), sharey=False)
if n_ct == 1:
    axes = [axes]

for ax, ct in zip(axes, all_ct_present):
    sub    = df[df['CellType'] == ct].copy()
    groups = [g for g in order if g in sub['Transition'].unique()]

    sns.boxplot(
        data=sub, x='Transition', y='p_other',
        hue='Transition', order=groups,
        palette=transition_colors, legend=False, ax=ax,
        flierprops=dict(marker='o', markersize=3, alpha=0.5),
        width=0.5,
    )
    sns.stripplot(
        data=sub, x='Transition', y='p_other',
        order=groups, color=cell_type_colors.get(ct, '#555555'),
        alpha=0.7, size=5, jitter=True, ax=ax,
    )

    ylim  = ax.get_ylim()
    y_top = ylim[1]

    # n labels
    for i, grp in enumerate(groups):
        n = len(sub[sub['Transition'] == grp])
        ax.text(i, y_top * 1.01, f'n={n}',
                ha='center', va='bottom', fontsize=8, fontweight='bold')

    # Pairwise t-test brackets for all group pairs present
    _pair_levels = [y_top * 1.12, y_top * 1.22, y_top * 1.32]
    _pair_defs   = [
        ('Stable_Ctrl', 'Stable'),
        ('Stable_Ctrl', 'Unstable'),
        ('Stable',      'Unstable'),
    ]
    _y_max = y_top
    _k = 0
    for (g1, g2) in _pair_defs:
        if g1 not in groups or g2 not in groups:
            continue
        v1 = sub[sub['Transition'] == g1]['p_other']
        v2 = sub[sub['Transition'] == g2]['p_other']
        if len(v1) < 2 or len(v2) < 2:
            continue
        _, pval = stats.ttest_ind(v2, v1)
        stars   = get_sig(pval)
        i1, i2  = groups.index(g1), groups.index(g2)
        y_lvl   = _pair_levels[_k % len(_pair_levels)]
        ax.plot([i1, i1, i2, i2],
                [y_lvl - y_top*0.02, y_lvl, y_lvl, y_lvl - y_top*0.02],
                color='#333333', lw=0.9)
        pval_txt = f'p={pval:.2e}' if pval < 0.001 else f'p={pval:.3f}'
        ax.text((i1+i2)/2, y_lvl + y_top*0.01,
                f'{stars}  {pval_txt}',
                ha='center', va='bottom', fontsize=6, fontweight='bold',
                color='black' if stars != 'ns' else '#888888')
        _y_max = max(_y_max, y_lvl)
        _k += 1
    if _k > 0:
        ax.set_ylim(ylim[0], _y_max * 1.12)

    ax.axhline(0.5, color='gray', lw=1, ls='--', alpha=0.5)
    ax.set_title(ct, fontsize=12, fontweight='bold',
                 color=cell_type_colors.get(ct, 'black'))
    ax.set_xlabel('')
    ax.set_ylabel('P(transition away)', fontsize=9)
    ax.tick_params(axis='x', labelsize=9)

plt.suptitle('P(transition away from own identity) = 1 − P(self)\nper cell type: Stable_Ctrl vs Stable vs Unstable',
             fontsize=13, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig('p_other_per_celltype2.pdf', dpi=300, bbox_inches='tight')
plt.show()
plt.close()
print("Saved: p_other_per_celltype2.pdf")

# ─────────────────────────────────────────────
# STATS SUMMARY TABLE
# ─────────────────────────────────────────────
summary_rows = []
for ct in all_ct_present:
    sub    = df[df['CellType'] == ct]
    groups = [g for g in order if g in sub['Transition'].unique()]
    for grp in groups:
        vals = sub[sub['Transition'] == grp]['p_other']
        row  = {
            'CellType':   ct,
            'Transition': grp,
            'N':          len(vals),
            'mean':       vals.mean(),
            'median':     vals.median(),
            'std':        vals.std(),
        }
        summary_rows.append(row)

    # Pairwise t-tests for all pairs present
    _stat_pairs = [
        ('Stable_Ctrl', 'Stable',   'ttest_Stable_vs_Ctrl_p'),
        ('Stable_Ctrl', 'Unstable', 'ttest_Unstable_vs_Ctrl_p'),
        ('Stable',      'Unstable', 'ttest_Unstable_vs_Stable_p'),
    ]
    for g1, g2, col_name in _stat_pairs:
        if g1 not in groups or g2 not in groups:
            continue
        v1 = sub[sub['Transition'] == g1]['p_other']
        v2 = sub[sub['Transition'] == g2]['p_other']
        if len(v1) > 1 and len(v2) > 1:
            _, pval = stats.ttest_ind(v2, v1)
            sig     = get_sig(pval)
            # Attach to the two relevant rows
            for row in summary_rows:
                if row['CellType'] == ct and row['Transition'] in (g1, g2):
                    row[col_name]              = pval
                    row[col_name + '_sig']     = sig

summary_df = pd.DataFrame(summary_rows).fillna('')
print("\nStats summary:")
print(summary_df.to_string(index=False))
summary_df.to_csv('p_other_stats_summary2.csv', index=False)
print("Saved: p_other_stats_summary.csv")
```
